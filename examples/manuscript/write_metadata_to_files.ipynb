```python
# get file sizes of mzml files in both folders
import matplotlib.pyplot as plt
import numpy as np
pos_raw_data_sizes = {f: os.path.getsize(os.path.join(raw_data_folder, f))/1e6 for f in raw_data_files if 'pos' in f.lower()}
pos_converted_data_sizes = {f: os.path.getsize(os.path.join(converted_data_folder, f))/1e6 for f in my_files if 'pos' in f.lower()}
neg_raw_data_sizes = {f: os.path.getsize(os.path.join(raw_data_folder, f))/1e6 for f in raw_data_files if 'neg' in f.lower()}
neg_converted_data_sizes = {f: os.path.getsize(os.path.join(converted_data_folder, f))/1e6 for f in my_files if 'neg' in f.lower()}
fig,ax = plt.subplots(1,2,figsize=(10,5), sharey=True, sharex=True)
# histogram of file sizes
edges = np.linspace(50,90,50)
ax[0].hist(list(pos_raw_data_sizes.values()), bins=edges, alpha=0.5, label='Original Data')
ax[0].hist(list(pos_converted_data_sizes.values()), bins=edges, alpha=0.5, label='Encoded Data')
ax[0].set_xlabel('File Size (MB)', fontsize=18)
ax[0].set_ylabel('Frequency', fontsize=18)
# ax[0].set_title('Positive Mode', fontsize=18)

# histogram of negative mode file sizes
ax[1].hist(list(neg_raw_data_sizes.values()), bins=edges, alpha=0.5, label='Original Data')
ax[1].hist(list(neg_converted_data_sizes.values()), bins=edges, alpha=0.5, label='Encoded Data')
ax[1].set_xlabel('File Size (MB)', fontsize=18)
# ax[1].set_ylabel('Frequency', fontsize=18)
# ax[1].set_title('Negative Mode', fontsize=18)
ax[1].legend(fontsize=16)

# add "a" and "b" labels to the subplots
ax[0].text(0.05, 0.9, 'a', transform=ax[0].transAxes, fontsize=33)
ax[1].text(0.05, 0.9, 'b', transform=ax[1].transAxes, fontsize=33)

# make the tick labels larger
ax[0].tick_params(axis='both', which='major', labelsize=16)
ax[1].tick_params(axis='both', which='major', labelsize=16)

# remove the top and right spines
ax[0].spines['top'].set_visible(False)
ax[0].spines['right'].set_visible(False)
ax[1].spines['top'].set_visible(False)
ax[1].spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```
